# Linear Regression Using Gradient Descent

حل التاسك بالكامل مع الشرح والتجربة والتنفيذ بدون استخدام sklearn.

##  Load & Understand the Data

- **X** تمثل مساحة البيت بالمتر المربع
- **y** تمثل سعر البيت (بالآلاف)

العلاقة خطية: كل زيادة في المساحة تقابلها زيادة ثابتة في السعر.

```python

import numpy as np
import matplotlib.pyplot as plt

X = np.array([50, 60, 70, 80, 90])
y = np.array([150, 180, 210, 240, 270])
```

##  Linear Regression Class (Gradient Descent)

```python

class LinearRegressionGD:
    def __init__(self, learning_rate=0.001, n_iters=100):
        self.lr = learning_rate
        self.n_iters = n_iters
        self.theta_0 = 0
        self.theta_1 = 0
        self.losses = []

    def fit(self, X, y):
        n = len(X)
        for _ in range(self.n_iters):
            y_pred = self.theta_0 + self.theta_1 * X
            error = y_pred - y
            
            d_theta0 = (2/n) * sum(error)
            d_theta1 = (2/n) * sum(error * X)
            
            self.theta_0 -= self.lr * d_theta0
            self.theta_1 -= self.lr * d_theta1
            
            sse = sum(error ** 2)
            self.losses.append(sse)

    def predict(self, X):
        return self.theta_0 + self.theta_1 * X

    def plot_training(self, X, y):
        plt.figure()
        plt.plot(self.losses)
        plt.xlabel("Iterations")
        plt.ylabel("SSE")
        plt.title("SSE over Iterations")
        plt.show()

        plt.figure()
        plt.scatter(X, y)
        plt.plot(X, self.predict(X))
        plt.xlabel("House Size")
        plt.ylabel("Price")
        plt.title("Regression Line")
        plt.show()
```

##  Create & Train the Model

```python

model = LinearRegressionGD(learning_rate=0.001, n_iters=100)
model.fit(X, y)

print("Theta 0 (Intercept):", model.theta_0)
print("Theta 1 (Slope):", model.theta_1)
```


###  What do theta_0 and theta_1 represent?
- **theta_0**: السعر الأساسي عندما تكون المساحة = 0
- **theta_1**: مقدار زيادة السعر لكل متر مربع إضافي


##  Prediction

```python

predicted_price = model.predict(70)
print("Predicted price for 70 m²:", predicted_price)
```


###  Is the prediction reasonable?
نعم، لأن 70 موجودة داخل نطاق الداتا والنتيجة قريبة جدًا من القيمة الحقيقية (210).


##  Visualization

```python

model.plot_training(X, y)
```

## Experimentation

```python

# Very large learning rate
model_large_lr = LinearRegressionGD(learning_rate=0.1, n_iters=50)
model_large_lr.fit(X, y)

# Very small learning rate
model_small_lr = LinearRegressionGD(learning_rate=0.00001, n_iters=50)
model_small_lr.fit(X, y)

print("Final SSE (Large LR):", model_large_lr.losses[-1])
print("Final SSE (Small LR):", model_small_lr.losses[-1])
```

##Bonus: MSE Method

```python

def calculate_mse(y_true, y_pred):
    return np.mean((y_true - y_pred) ** 2)

print("MSE:", calculate_mse(y, model.predict(X)))
```
