```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("bank.csv")
```

```python
print(df.head())
```

```python
print(df.info())
```

```python
print(df.isnull().sum())
```

```python
df_clean = df.copy()

for col in df_clean.columns:
    if df_clean[col].dtype != 'object':
        df_clean[col] = df_clean[col].fillna(df_clean[col].median())
    else:
        df_clean[col] = df_clean[col].fillna(df_clean[col].mode()[0])
```

```python
print(df_clean.isnull().sum())
```

```python
num_cols = df_clean.select_dtypes(include='number').columns.tolist()
if num_cols:
    plt.figure()
    df_clean[num_cols[0]].hist()
    plt.title(f"Distribution of {num_cols[0]}")
    plt.xlabel(num_cols[0])
    plt.ylabel("Frequency")
    plt.show()
```

```python
cat_cols = df_clean.select_dtypes(include='object').columns.tolist()
if cat_cols:
    plt.figure()
    df_clean[cat_cols[0]].value_counts().plot(kind="bar")
    plt.title(f"Count of {cat_cols[0]}")
    plt.xlabel(cat_cols[0])
    plt.ylabel("Count")
    plt.show()
```

```python
plt.figure(figsize=(6,4))
df_clean['age'].hist()
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(6,4))
df_clean['balance'].hist()
plt.title("Balance Distribution")
plt.xlabel("Balance")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(6,4))
df_clean['deposit'].value_counts().plot(kind='bar')
plt.title("Deposit Result")
plt.xlabel("Deposit")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(x='deposit', y='age', data=df_clean)
plt.title("Age vs Deposit")
plt.show()
```

```python
num_cols = df_clean.select_dtypes(include='number')
plt.figure(figsize=(8,5))
sns.heatmap(num_cols.corr(), annot=True)
plt.title("Correlation Heatmap")
plt.show()
```

```python
plt.figure(figsize=(6,4))
df_clean['job'].value_counts().head(10).plot(kind='bar')
plt.title("Top 10 Jobs")
plt.xlabel("Job")
plt.ylabel("Count")
plt.show()
```
