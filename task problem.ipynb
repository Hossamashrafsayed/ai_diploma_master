```python
def searchInsert(nums, target):
    left, right = 0, len(nums) - 1
    
    while left <= right:
        mid = (left + right) // 2
        
        if nums[mid] == target:
            return mid
        elif nums[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    
    return left
```

```python
def calculate(s: str) -> int:
    s = s.replace(" ", "")
    stack = []
    num = 0
    sign = '+'
    
    for i, ch in enumerate(s):
        if ch.isdigit():
            num = num * 10 + int(ch)
        
        if not ch.isdigit() or i == len(s) - 1:
            if sign == '+':
                stack.append(num)
            elif sign == '-':
                stack.append(-num)
            elif sign == '*':
                stack.append(stack.pop() * num)
            elif sign == '/':
                prev = stack.pop()
                stack.append(int(prev / num))  # truncate toward zero
            
            sign = ch
            num = 0
    
    return sum(stack)
```
